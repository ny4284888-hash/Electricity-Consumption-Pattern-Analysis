```python
data = {
    "Date": ["2026-10-01", "2026-10-02", "2026-10-03", "2026-10-04", "2026-10-05"],
    "Electricity_Consumption": [12, 15, 18, 10, 20]
}

df = pd.DataFrame(data)

df
```

```python
print("Total Electricity Consumption:", df["Electricity_Consumption"].sum(), "kWh")

print("Average Daily Consumption:", df["Electricity_Consumption"].mean(), "kWh")
```

```python
highest = df["Electricity_Consumption"].max()

print("Highest Consumption:", highest, "kWh")
```

```python
highest_day = df.loc[df["Electricity_Consumption"].idxmax()]

print("Highest Consumption Date:", highest_day["Date"])
print("Electricity Used:", highest_day["Electricity_Consumption"], "kWh")
```

```python
plt.figure(figsize=(8, 4))

plt.plot(df["Date"], df["Electricity_Consumption"], marker="o")

plt.title("Daily Electricity Consumption")
plt.xlabel("Date")
plt.ylabel("Electricity Consumption (kWh)")

plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
df["Date"] = pd.to_datetime(df["Date"])

print(df.dtypes)
```

```python
df["Day"] = df["Date"].dt.day_name()

df["Day_Type"] = df["Date"].dt.dayofweek.map(
    lambda x: "Weekend" if x >= 5 else "Weekday"
)

df
```

```python
day_comparison = df.groupby("Day_Type")["Electricity_Consumption"].mean()

print(day_comparison)
```

```python
day_comparison.plot(kind="bar", figsize=(6, 4))

plt.title("Weekday vs Weekend Electricity Consumption")
plt.xlabel("Day Type")
plt.ylabel("Average Consumption (kWh)")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.show()
```

```python
Q1 = df["Electricity_Consumption"].quantile(0.25)
Q3 = df["Electricity_Consumption"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = df[
    (df["Electricity_Consumption"] < lower_limit) |
    (df["Electricity_Consumption"] > upper_limit)
]

print("Unusual Readings:")
print(outliers)
```

```python
plt.figure(figsize=(8, 4))

plt.boxplot(df["Electricity_Consumption"])

plt.title("Electricity Consumption - Outlier Detection")
plt.ylabel("Consumption (kWh)")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8, 4))

plt.bar(df["Date"].dt.strftime("%d-%b"), df["Electricity_Consumption"])

plt.title("Daily Electricity Consumption Analysis")
plt.xlabel("Date")
plt.ylabel("Consumption (kWh)")

plt.grid(axis="y")
plt.show()
```

```python
dates = pd.date_range(start="2026-10-01", periods=30, freq="D")

consumption = [
    12, 15, 18, 10, 20, 16, 14, 19, 22, 17,
    13, 21, 18, 15, 24, 20, 16, 14, 23, 19,
    17, 25, 21, 18, 15, 22, 20, 16, 24, 19
]

data = {
    "Date": dates,
    "Electricity_Consumption": consumption
}

df = pd.DataFrame(data)

df.head()
```

```python
print("Total Records:", len(df))
print(df.tail())
```

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Day"] = df["Date"].dt.day_name()

df["Day_Type"] = df["Date"].dt.dayofweek.map(
    lambda x: "Weekend" if x >= 5 else "Weekday"
)

df.head(10)
```

```python
df["Day_Type"] = df["Date"].dt.day_of_week.map(
    lambda x: "Weekend" if x >= 5 else "Weekday"
)
```

```python
day_comparison = df.groupby("Day_Type")["Electricity_Consumption"].mean()

print("Average Electricity Consumption:")
print(day_comparison)
```

```python
day_comparison.plot(kind="bar", figsize=(7, 4))

plt.title("Weekday vs Weekend Electricity Consumption")
plt.xlabel("Day Type")
plt.ylabel("Average Consumption (kWh)")
plt.xticks(rotation=0)
plt.grid(axis="y")

plt.show()
```

```python
highest_day = df.loc[df["Electricity_Consumption"].idxmax()]
lowest_day = df.loc[df["Electricity_Consumption"].idxmin()]

print("Highest Consumption:")
print("Date:", highest_day["Date"].date())
print("Consumption:", highest_day["Electricity_Consumption"], "kWh")

print("\nLowest Consumption:")
print("Date:", lowest_day["Date"].date())
print("Consumption:", lowest_day["Electricity_Consumption"], "kWh")
```

```python
extreme_days = pd.DataFrame({
    "Category": ["Highest", "Lowest"],
    "Consumption": [
        highest_day["Electricity_Consumption"],
        lowest_day["Electricity_Consumption"]
    ]
})

plt.figure(figsize=(6, 4))
plt.bar(extreme_days["Category"], extreme_days["Consumption"])

plt.title("Highest vs Lowest Electricity Consumption")
plt.xlabel("Category")
plt.ylabel("Consumption (kWh)")
plt.grid(axis="y")

plt.show()
```

```python
Q1 = df["Electricity_Consumption"].quantile(0.25)
Q3 = df["Electricity_Consumption"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = df[
    (df["Electricity_Consumption"] < lower_limit) |
    (df["Electricity_Consumption"] > upper_limit)
]

print("Unusual Electricity Readings:")
print(outliers)
```

```python
plt.figure(figsize=(7, 4))

plt.boxplot(df["Electricity_Consumption"])

plt.title("Electricity Consumption - Outlier Detection")
plt.ylabel("Consumption (kWh)")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df["Date"],
    df["Electricity_Consumption"],
    marker="o"
)

plt.title("30 Days Electricity Consumption Analysis")
plt.xlabel("Date")
plt.ylabel("Consumption (kWh)")
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Creating a copy of original dataset
df_cleaned = df.copy()

# Checking missing values
print("Missing Values Before Cleaning:")
print(df_cleaned.isnull().sum())

# Filling missing consumption values with average
df_cleaned["Electricity_Consumption"] = (
    df_cleaned["Electricity_Consumption"].fillna(
        df_cleaned["Electricity_Consumption"].mean()
    )
)

print("\nMissing Values After Cleaning:")
print(df_cleaned.isnull().sum())
```

```python
print("Total Duplicate Rows:")
print(df_cleaned.duplicated().sum())

print("\nDuplicate Records:")
print(df_cleaned[df_cleaned.duplicated()])
```

```python
print("First Date:", df_cleaned["Date"].min())
print("Last Date:", df_cleaned["Date"].max())

print("\nTotal Records:", len(df_cleaned))
print("Unique Dates:", df_cleaned["Date"].nunique())

print("\nDate Validation:")
print("Duplicate Dates:", df_cleaned["Date"].duplicated().sum())
```

```python
expected_dates = pd.date_range(
    start=df_cleaned["Date"].min(),
    end=df_cleaned["Date"].max(),
    freq="D"
)

missing_dates = expected_dates.difference(df_cleaned["Date"])

print("Missing Dates:")
print(missing_dates)

print("\nTotal Missing Dates:", len(missing_dates))
```

```python
invalid_readings = df_cleaned[
    df_cleaned["Electricity_Consumption"] <= 0
]

print("Invalid Electricity Readings:")
print(invalid_readings)

print("\nTotal Invalid Readings:", len(invalid_readings))
```

```python
hourly_data = {
    "Hour": list(range(24)),
    "Electricity_Consumption": [
        0.3, 0.2, 0.2, 0.2, 0.3, 0.5,
        0.8, 1.2, 1.5, 1.0, 0.8, 0.7,
        0.8, 0.9, 0.8, 0.7, 0.9, 1.2,
        1.8, 2.0, 1.7, 1.3, 0.9, 0.5
    ]
}

hourly_df = pd.DataFrame(hourly_data)

hourly_df
```

```python
def get_time_period(hour):
    if 0 <= hour <= 5:
        return "Night"
    elif 6 <= hour <= 11:
        return "Morning"
    elif 12 <= hour <= 16:
        return "Afternoon"
    else:
        return "Evening"

hourly_df["Time_Period"] = hourly_df["Hour"].apply(get_time_period)

hourly_df
```

```python
period_consumption = hourly_df.groupby("Time_Period")["Electricity_Consumption"].sum()

print("Electricity Consumption by Time Period:")
print(period_consumption)
```

```python
period_consumption.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Electricity Consumption by Time Period")
plt.xlabel("Time Period")
plt.ylabel("Total Consumption (kWh)")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.show()
```

```python
peak_hour = hourly_df.loc[
    hourly_df["Electricity_Consumption"].idxmax()
]

print("Peak Consumption Hour:", peak_hour["Hour"])
print("Highest Consumption:", peak_hour["Electricity_Consumption"], "kWh")
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    hourly_df["Hour"],
    hourly_df["Electricity_Consumption"],
    marker="o"
)

plt.title("Hourly Electricity Consumption Pattern")
plt.xlabel("Hour of the Day")
plt.ylabel("Electricity Consumption (kWh)")
plt.xticks(range(24))
plt.grid(True)
plt.show()
```

```python
print("Hourly Dataset Summary")
print("-----------------------")

print("Total Consumption:", hourly_df["Electricity_Consumption"].sum(), "kWh")
print("Average Consumption:", hourly_df["Electricity_Consumption"].mean(), "kWh")
print("Maximum Consumption:", hourly_df["Electricity_Consumption"].max(), "kWh")
print("Minimum Consumption:", hourly_df["Electricity_Consumption"].min(), "kWh")
print("Total Hours Analyzed:", len(hourly_df))
```
